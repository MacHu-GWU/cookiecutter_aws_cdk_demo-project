This page demonstrates how a docs page written as a Jupyter notebook mixes three kinds of cells. Open `docs/source/02-Sample-Jupyter-Notebook-Document/index.ipynb` next to it and copy what you need.

- **Markdown cells** for prose, lists, tables, images, and links.
- **Code cells** for runnable examples. Their outputs are committed with the notebook.
- **Raw reStructuredText cells** (Raw NBConvert Format: reStructuredText) for everything Sphinx-specific: the labeled page title above, cross-references, and directives.

The reStructuredText version of this page is [Sample RestructuredText Document](../01-Sample-RestructuredText-Document/index.rst).
